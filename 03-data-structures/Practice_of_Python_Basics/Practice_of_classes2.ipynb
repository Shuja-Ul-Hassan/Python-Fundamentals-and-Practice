# Classes
A class in Python OOP is like a blueprint for making objects.

It tells Python what data an object should have (Attributes)

And what actions it can do (Methods)

## Example:
Class = Recipe (instructions for making a cake)

Object = Actual Cake (made using the recipe)

So, a class is just the recipe, and objects are the real things you bake from it.

```python
class Dog:  # Defines a class named Dog

    def __init__(self, name, breed):  
        # __init__ is a special constructor method.
        # It runs automatically when a new Dog object is created.
        # 'self' refers to the specific Dog object being created.
        # 'name' and 'breed' are parameters provided by the user.

        self.name = name # Assigns the given name to this Dog object
        self.breed = breed # Assigns the given breed to this Dog object

    def bark(self):# Defines a method (function inside the class)
        print(f"{self.name} barks!")  
        # Uses f-string to print the dog's name followed by "barks!"

# Create an object (instance) of Dog
my_dog = Dog("Bozo", "Golden Retriever")  
# This calls __init__, passing "Buddy" and "Golden Retriever"

#accessing the attributes
print(my_dog.name)# Accesses the 'name' attribute

#calling methods
my_dog.bark()# Calls the bark method 
```

```python
#Restaurant
class Restaurants:
    def __init__(self,name,cuisine_type):
        self.name=name;self.cuisine_type=cuisine_type;self.number_served=0

    def describe_restaurant(self):
        return f"The Name Of The Restaurant is {self.name}\nThe Cuisine Type Of This Restaurant is {self.cuisine_type}"

    def open_restaurant(self):
        return f"The {self.name} is open!"
    
    def set_number_served(self):
        self.number_served=int(input("Enter the number of customers that have been served: "))

    def increment_number_served(self,increment):
        self.number_served+=increment

restaurant=Restaurants("Si Wok","Chinese")
restaurant1=Restaurants("AlKhan","Traditional")
restaurant2=Restaurants("Maalga","Traditional")
restaurant3=Restaurants("StoneFire","FastFood")
print(restaurant1.name)
print(f"{restaurant1.cuisine_type}\n")

restaurant.set_number_served()
print(f"The Restaurant {restaurant.name} has Served {restaurant.number_served} Customers\n")

restaurant.increment_number_served(6)
print(f"The Restaurant {restaurant.name} has now Served {restaurant.number_served} Customers\n")
print(f"{"-"*5}Restaurants Description{"-"*5}")
print(restaurant1.describe_restaurant())
print(f"\n{restaurant2.describe_restaurant()}\n")
print(f"{restaurant3.describe_restaurant()}")
print(restaurant1.open_restaurant())
```

```python
class User:
    def __init__(self,first_name,last_name,email,password,login_attempts):
        self.first_name=first_name
        self.last_name=last_name
        self.email=email
        self.password=password
        self.login_attempts=login_attempts

    def increment_login_attempts(self):
        self.login_attempts+=1
    def reset_login_attempts(self):
        self.login_attempts=0
    def describe_user(self):
        return f"First Name: {self.first_name}\nLast Name: {self.last_name}\nEmail ID: {self.email}\nPassword: {self.password}\n"
    def greet_user(self):
        return f"Hi {self.first_name}!"
    
user1=User("Shuja","Khan","Shuja1234@gmail.com","zx920f@f",1)
user2=User("Maria","Chunza","Maria1234@gmail.com","xz29f0f@",2)
user3=User("Alex","Fredrick","Alex1234@gmail.com","pinza3425d",3)

print(f"User1_Name: {user1.first_name} {user1.last_name}\nUser2_Email: {user2.email}\nUser3_Password: {user3.password}\n")
print(f"{"-"*5}Users Description{"-"*5}\n{user1.describe_user()}\n{user2.describe_user()}\n{user3.describe_user()}\n")
print(f"{"-"*5}Greeting Users{"-"*5}\n{user1.greet_user()}\n{user2.greet_user()}\n{user3.greet_user()}")

user1.increment_login_attempts()
user1.increment_login_attempts()
user1.increment_login_attempts()

print(f"\n{"-"*5}Login Attempts{"-"*5}\nThe Number of login Attempts For User 1 are: {user1.login_attempts}")

user1.reset_login_attempts()

print(f"The Number of login Attempts Now For User 1 are: {user1.login_attempts}")
```

# Inheritance

~ SUPERCLASS: 

    A superclass (also called a parent class or base class) is the class you inherit from.

~ SUBCLASS: 

    A subclass (also called a child class or derived class) is the class that inherits from the superclass.

    It can use the superclass’s methods/attributes and also define its own.

~ SUPER().__INIT__(): 

    The super() function is used to call methods from the superclass inside the subclass.

    Without it, you’d have to rewrite the code for setting self.name in every subclass.


#### Superclass = parent class (common blueprint).

#### Subclass = child class (inherits and extends).

#### super().__init__() = lets the subclass reuse the parent’s constructor instead of rewriting it.

```python
class IceCreamStand(Restaurants):
    def __init__(self,name,cuisine_type,flavours):
        self.flavours=flavours
        super().__init__(name,cuisine_type)
    def show_flavours(self):
        for x in self.flavours:
            print(x)
Customer1=IceCreamStand("zico","FrozenTreats",["chocolate","Vanilla","Strawberry"])
Customer1.show_flavours()
```
