```python
class Cat:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def meows(self):
        print(f"{self.name} meows!")
        
    def breeds(self):
        print(f"My Cat Is {self.breed} Breed")
    
class Kitten(Cat):
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def plays(self):
        print(f"My {self.name} Plays!")
        
    def aged(self):
        print(f"My {self.name} is {self.age} old ")
        
my_kitten=Kitten("CupCake","3 Months")

my_cat = Cat("Snuggles", "Persian")

print(my_cat.name)
my_cat.meows() 
my_cat.breeds()

print(my_kitten.name)
my_kitten.plays()
my_kitten.aged()
```

```python
class Person:
    def __init__(self,name,age,city):
        self.name=name
        self.age=age
        self.city=city
    def Name(self):      
        print(f"My Name is {self.name}")
    def Age(self):
        print(f"My Age is {self.age}")
    def City(self):
        print(f"I live in {self.city}")
theperson=Person("Shuja",18,"Lahore")
theperson.Name()
theperson.City()
theperson.Age()
```

```python
class Car:
    def __init__(self,make,model,year):
        self.make=make
        self.model=model
        self.year=year
    def Make(self):      
        print(f"My Car is {self.make}")
    def Model(self):
        print(f"My car is {self.make} {self.model}")
    def Year(self):
        print(f"I bought it in the year {self.year}")
thecar=Car("Corolla","GLI",2018)
thecar.Make()
thecar.Model()
thecar.Year()
```

```python
class Circle:
    def __init__(self, radius):
        self.radius= radius
        self.pi = 3.14159
    def calculate_area(self):
        Area= self.pi * (self.radius ** 2)
        print(f"The Area is {Area:.2f}")
    def calculate_circumference(self):
        Circumference= 2 * self.pi * self.radius
        print(f"The Circumference is {Circumference:.2f}")
thesolution=Circle(23)
thesolution.calculate_area()
thesolution.calculate_circumference()
```

```python
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width

    def calculate_area(self):
        area=self.length*self.width
        print(f"The area of Rectangle is: {area}")

    def calculate_perimeter(self):
        perimeter=2*(self.length+self.width)
        print(f"The perimeter of Rectangle is: {perimeter}")
        
thesolutions=Rectangle(4,2)
thesolutions.calculate_area()
thesolutions.calculate_perimeter()
```

```python
class Book:
    def __init__(self, title, author, publication_year):
        self.title = title
        self.author = author
        self.publication_year = publication_year

    def thetitle(self):
        print(f"Title: {self.title}")
    def theauthor(self):
        print(f"Author: {self.author}")
    def theyear(self):
        print(f"Year: {self.publication_year}")
        
thedetails=Book("The Stranger","Albert Camus",1942)
thedetails.thetitle()
thedetails.theauthor()
thedetails.theyear()
```

```python
class Shape:
    def calculate_area(self):
        print("This shape does not have a specific area calculation defined yet.")

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
        self.pi = 3.14159
    def calculate_area(self):
        area = self.pi*(self.radius ** 2)
        print(f"Circle Area: {area:.2f}")
thecircle=Circle(4)

class Rectangle(Shape):
    def __init__(self,length,width):
        self.length=length
        self.width=width
    def calculate_area(self):
        area=self.length*self.width
        print(f"Rectangle Area: {area}")
therectangle=Rectangle(2,4)

class Triangle(Shape):
    def __init__(self,base,height):
        self.base=base
        self.height=height
    def calculate_area(self):
        area=0.5*self.base*self.height
        print(f"Triangle Area: {area}")
thetriangle=Triangle(5,3)

thecircle.calculate_area()
therectangle.calculate_area()
thetriangle.calculate_area()
```

```python
class Animal:
    def __init__(self,name,sound="Generic sound"):
        self.name = name
        self.sound = sound
    def make_sound(self):
        print(f"{self.name} says '{self.sound}'")

class Dog(Animal):
    def __init__(self, name):
        self.name=name
        self.sound="Woof!"
    def make_sound(self):
        print(f"{self.name} barks:'{self.sound}'")

class Cat(Animal):
    def __init__(self, name):
        self.name=name
        self.sound="Meow!"
        
class Cow(Animal):
    def __init__(self,name):
        self.name=name
        self.sound="Moo!"

my_dog=Dog("Buddy")
my_cat=Cat("Whiskers")
my_cow=Cow("Bessie")

my_dog.make_sound()
my_cat.make_sound()
my_cow.make_sound()

```
