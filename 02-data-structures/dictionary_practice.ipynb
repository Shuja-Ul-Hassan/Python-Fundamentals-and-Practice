```python
''' f) Create a dictionary to store product information: name, quantity, price.
Implement functions to:
Add new products
Update product quantities
Calculate total inventory value
Generate a report of low-stock items '''
inventory = {}

inventory['pencils'] = {'quantity': 600, 'price': 10}
inventory['eraser'] = {'quantity': 500, 'price': 10}
inventory['sharpner'] = {'quantity': 600, 'price': 10}

inventory['eraser']['quantity'] += 100  

total_value = sum(inventory[item]['quantity'] * inventory[item]['price'] for item in inventory)

print("Total Inventory Value:", total_value)

threshold=100
low_stock={item: data for item, data in inventory.items() if data['quantity']<=threshold}

print("Low Stock Report:")
for item, data in low_stock.items():
    print(f"- {item}: {data['quantity']} units")


```

```python
''' g) Create a dictionary to store student information: name, grades for different subjects.
Calculate the average grade for each student.
Find the highest and lowest grades in a specific subject.
Sort students by their average grades.'''


student_grades = {'Ali': {'Math': 85, 'English': 78, 'Science': 92},'Sara': {'Math': 90, 'English': 88, 'Science': 95},'Bilal': {'Math': 70, 'English': 65, 'Science': 80},'Zara': {'Math': 95, 'English': 90, 'Science': 85}}
average_grades = {}

for name in student_grades:
    subjects = student_grades[name]
    total = sum(subjects.values())
    count = len(subjects)
    average = total / count
    average_grades[name] = round(average, 2)

print(average_grades)

subject = 'Math'
subject_scores = {}
for name in student_grades:
    subject_scores[name] = student_grades[name][subject]

highest = max(subject_scores, key=subject_scores.get)
lowest = min(subject_scores, key=subject_scores.get)

print(f"Highest in {subject}: {highest} ({subject_scores[highest]})")
print(f"Lowest in {subject}: {lowest} ({subject_scores[lowest]})")

average_grades = {}
for name in student_grades:
    total = sum(student_grades[name].values())
    count = len(student_grades[name])
    average = total / count
    average_grades[name] = round(average, 2)

sorted_students = sorted(average_grades.items(), key=lambda x: x[1], reverse=True)

print("Students Sorted by Average Grade:")
for name, avg in sorted_students:
    print(name,":", avg)
```
