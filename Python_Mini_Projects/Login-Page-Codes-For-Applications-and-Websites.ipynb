This Code Removes The Spaces Between The Name And Converts It Into Lower Characters

```python
#Question no.1
name=input("Enter your Full Name:")
username=name.replace(" ","")
print(username.lower())
```

This Code Checks If The Email Entered Is Google Mail Or Not 

```python
# Question no.2
email=input("Enter Your Email:")
if "@gmail.com" in email:
    print("Valid Gmail")
else:
    print("Not Valid")
```

This Code Hides the Phones Number And Shows The Last 4 Digits

```python
# Question no.3
phonenumber=input("Enter Your Phone Number")
hiding="*"*(len(phonenumber)-4)
hiding+=phonenumber[-4:]
print(hiding)
```

This Code Replaces Certain Negative Words With Softer Alternatives.

```python
# Question no.4
statement=input("Enter a Statement:")
check=statement.lower()
check=check.replace("bad","not good")
check=check.replace("stupid","silly")
check=check.replace("dumb","unknowing")
print(statement.title())
print(check.title())
```

This Code Takes User's FullName And Gives The FirstName As Output

```python
# Question no.5
fullname=input("Enter Your Complete Name:")
first_name=fullname.strip()
first_name=fullname.split()[0]
print("Your first name is:", first_name)
```
