```python
import random
```

```python
def number_guessing_game():
  Answer=random.randrange(1,50)
  total_chances=0
  while total_chances<7:
    try:
      Guess=int(input("Guess a number between 1 to 50: "))
    except ValueError:
      print("Enter a Number please!")
    if Guess == Answer:
      print("Congratulations You Guessed It!")
      break
    elif Guess < 1 or Guess > 50:
      print("Enter a number between 1 to 50")
    elif Guess < Answer:
      print("Go Higher!")
    elif Guess > Answer:
      print("Go Lower!")
    total_chances+=1
    print(f"You Took {total_chances} Chances \nTotal Chances are 7")
number_guessing_game()
```
